```python
from Bio import SeqIO
from Bio import pairwise2
from Bio.pairwise2 import align, format_alignment
from collections import Counter
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction, molecular_weight as mw, MeltingTemp as mt
from Levenshtein import hamming, distance
import matplotlib.pyplot as plt
```

```python
# I. Sequence Manipulation

# 1. Find and display the total length of Sequence A and B.
from Bio import SeqIO
seqA = SeqIO.read("a_exercise.fasta", "fasta").seq
seqB = SeqIO.read("b_exercise.fasta", "fasta").seq

lenA = len(seqA)
lenB = len(seqB)
totalLen = lenA + lenB

print(f"total length of Sequence A and B = {totalLen}");

# 2. Find and display which index the first time codon (TAC) appears in both Sequence A and B.
print(f"Find and display which index the first time codon (TAC) appears in both Sequence A = {seqA.index('TAC')}");
print(f"Find and display which index the first time codon (TAC) appears in both Sequence B = {seqB.index('TAC')}");

# 3. Take the first 10 nucleotides from Sequence A, and the last 15 nucleotides from Sequence B. 
# Then combine the sequences and reverse it to a new sequence, Sequence C. 

# sequence[begins at:ends after:skips per]

firstTen = seqA[:10] # or seqA[0:10:1]
lastFifteen = seqB[-15:] # or seqB[-15::1]
combine = firstTen + lastFifteen
seqC = combine[::-1] # or seqC = combine[-1::-1] , ([-1::-1] from the last element, takes 1 steps backwards which means reverse)
print(f"result = {seqC}")

```

```python
# II. Sequence Analysis & Plotting

# 1. Determine and display the nucleotides base (A, C, G, T) frequency for Sequence A, B, and C with a chart using matplotlib library.

from collections import Counter
import matplotlib.pyplot as plt

countA = Counter(seqA)
countB = Counter(seqB)
countC = Counter(seqC)
# check
print(f"A = {countA} B = {countB} C = {countC}")

plt.bar(countA.keys(), countA.values())
plt.title("Sequence A ATGC Frequencies")
plt.show()

plt.bar(countB.keys(), countB.values())
plt.title("Sequence B ATGC Frequencies")
plt.show()

plt.bar(countC.keys(), countC.values())
plt.title("Sequence C ATGC Frequencies")
plt.show()
```

```python
# 2. Determine and display the GC & AT content percentage for Sequence A, B, and C

from Bio.SeqUtils import gc_fraction

# gc_fraction = to show the GC contents in a sequence
gcA = gc_fraction(seqA)*100
print(f"GC content of Sequence A = {gcA:.2f}%")
print(f"AT content of Sequence A = {100-(gcA):.2f}%")

gcB = gc_fraction(seqB)*100
print(f"GC content of Sequence B = {gcB:.2f}%")
print(f"AT content of Sequence A = {100 - (gcB):.2f}%")

gcC = gc_fraction(seqC)*100
print(f"GC content of Sequence C = {gcC:.2f}%")
print(f"AT content of Sequence C = {100-(gcC):.2f}%")

# 3. Determine and display the melting temperature for Sequence A, B, and C (using built-inWallace Method)

from Bio.SeqUtils import MeltingTemp as mt
mTempA = mt.Tm_Wallace(seqA)
mTempB = mt.Tm_Wallace(seqB)
mTempC = mt.Tm_Wallace(seqC)
print(f"the melting temperature for Sequence A = {mTempA} and B = {mTempB}")

# 4. Determine and display the molecular weight for Sequence A, B and C.

from Bio.SeqUtils import molecular_weight as mw
mWeightA = mw(seqA)
mWeightB = mw(seqB)
mWeightC = mw(seqC)
print(f"the molecular weight for Sequence B = {mWeightB:.2f} and C = {mWeightC:.2f}")
```

```python
# III. DNA & mRNA Protein Synthesis

# 1. Transcribe and display the DNA Sequence A, B, and C into mRNA Sequences

# transcribe() = Replaces thymine (T) with uracil (U)
mRNAa = seqA.transcribe()
mRNAb = seqB.transcribe()
mRNAc = seqC.transcribe()
print(f"mRNA_A = {mRNAa}")
print(f"mRNA_B = {mRNAb}")
print(f"mRNA_C = {mRNAc}")

# 2. Translate and display the mRNA Sequence A, B, and C into amino acids sequences

# translate() = translates mRNA sequence into amino acids, with using letters as sign
# ex: Q = Glutamine, L = Leucine, K = Lysine, etc.
aminoAc_A = mRNAa.translate()
aminoAc_B = mRNAb.translate()
aminoAc_C = mRNAc.translate()
print(f"amino_ac_A = {aminoAc_A}")
print(f"amino_ac_B = {aminoAc_B}")
print(f"amino_ac_C = {aminoAc_C}")
```

```python
# IV. Sequence Alignment & Similarities

# Global vs Local:
# Global = Aligns entire sequences from beginning to end
# Local = Only shows on the best‑matching substring(s)

# Global Alignment:
# ACTGCTACGCTA
#     ||||||
# ----CTACGC--

# Local Alignment:
# CTACGC
# ||||||
# CTACGC

# 1. Perform local alignment pairwise between Sequence A and Sequence “CGCTACTAGCATGCGCAT” and display the alignments’ score
from Bio.pairwise2 import format_alignment, align
localAlignA = align.localxx(seqA, "CGCTACTAGCATGCGCAT")
for alignment in localAlignA:
    print(format_alignment(*alignment))

# for loop = to iterate over all possible local alignments. Without the loop, you'd only get the first one. The loop is the correct way to see all results.

# 2. Perform global alignment pairwise between Sequence B and Sequence “CAGTGCGCTAGCAGCTACGTGTG” and display the alignments’ score
from Bio import pairwise2
globalAlignB = pairwise2.align.globalxx(seqB, "CAGTGCGCTAGCAGCTACGTGTG")
for alignment in globalAlignB:
    print(format_alignment(*alignment))

# 3. Find and display the Levenshtein distance between Sequence A and B
from Levenshtein import distance
print (f"Dsitance between A and B: {distance(seqA, seqB)}")
```

```python
# V. Distance
seq1 = "ATCGATCG"
seq2 = "ATCAATCG"

# 1. Levenshtein Distance
from Levenshtein import distance
print(distance(seq1, seq2))
print(distance(seqA, seqB))

# 2. Hamming Distance
from Levenshtein import hamming
print(hamming(seq1, seq2))
print(hamming(seqA, seqB))
```
