```python
# Import Library
from Bio import SeqIO
```

```python
# Read file GenBank
record = SeqIO.read("sequence.gb", "genbank")
print(record.id)
print(record.name)
print(record.description)
print(record.seq)
```

```python
# Read FASTA
seqA = SeqIO.read("a.fasta", "fasta").seq
seqB = SeqIO.read("b.fasta", "fasta").seq

# print(f"Sequnece: {seq}")
print(f"Sequence A: {seqA}")
print(f"Sequence B: {seqB}")
```

```python
# 1) Sequence Manipulation
# Find and display the total length of Sequence A and B. 
len_SeqA = len(seqA)
len_SeqB = len(seqB)
total_len = len_SeqA + len_SeqB
print(f"Total length of Sequence A and B: {total_len}")

# Find and display which index the first time codon (TAC) appears in both Sequence A and B.
print(f"Index of first time codon \"TAC\" in Sequence A: {seqA.find('TAC')}")
print(f"Index of first time codon \"TAC\" in Sequence B: {seqB.find('TAC')}")

# Take the first 10 nucleotides from Sequence A, and the last 15 nucleotides from Sequence B. Then combine the sequences and reverse it to a new sequence, Sequence C.
ten_first = seqA[:10]
last_fifteen = seqB[-15:]
combine = ten_first + last_fifteen
seqC = combine[::-1]    # Reverse the sequence
print(f"Sequence C: {seqC}")
```

```python
# 2) Sequence Analysis & Plotting
# Determine and display the nucleotides base (A, C, G, T) frequency for Sequence A, B, and C with a chart using matplotlib library. 
from collections import Counter
import matplotlib.pyplot as plt

counterA = Counter(seqA)
counterB = Counter(seqB)
counterC = Counter(seqC)
print(counterA)
print(counterB)
print(counterC)

plt.bar(counterA.keys(), counterA.values())
plt.title("GCAT Frequency in Sequence A")
plt.show()
plt.bar(counterB.keys(), counterB.values())
plt.title("GCAT Frequency in Sequence B")
plt.show()
plt.bar(counterC.keys(), counterC.values())
plt.title("GCAT Frequency in Sequence C")
plt.show()

# Determine and display the GC & AT content percentage for Sequence A, B, and C.
from Bio.SeqUtils import gc_fraction
gc_seqA = gc_fraction(seqA) * 100
print(f"GC Content Percentage for Sequence A: {gc_seqA:.2f}%")
print(f"AT Content Percentage for Sequence A: {100-gc_seqA:.2f}%")
gc_seqB = gc_fraction(seqB) * 100
print(f"GC Content Percentage for Sequence B: {gc_seqB:.2f}%")
print(f"AT Content Percentage for Sequence B: {100-gc_seqB:.2f}%")
gc_seqC = gc_fraction(seqC) * 100
print(f"GC Content Percentage for Sequence C: {gc_seqC:.2f}%")
print(f"AT Content Percentage for Sequence C: {100-gc_seqC:.2f}%\n")

# Determine and display the melting temperature for Sequence A, B, and C (using built-inWallace Method).
from Bio.SeqUtils import MeltingTemp as mt
mt_seqA = mt.Tm_Wallace(seqA)
print(f"Melting Temperature for Sequence A: {mt_seqA:.2f}°C")
mt_seqB = mt.Tm_Wallace(seqB)
print(f"Melting Temperature for Sequence B: {mt_seqB:.2f}°C")
mt_seqC = mt.Tm_Wallace(seqC)
print(f"Melting Temperature for Sequence C: {mt_seqC:.2f}°C\n")

# Determine and display the molecular weight for Sequence A, B, and C.
from Bio.SeqUtils import molecular_weight
mw_seqA = molecular_weight(seqA)
print(f"Molecular Weight for Sequence A: {mw_seqA:.2f} g/mol")
mw_seqB = molecular_weight(seqB)
print(f"Molecular Weight for Sequence B: {mw_seqB:.2f} g/mol")
mw_seqC = molecular_weight(seqC)
print(f"Molecular Weight for Sequence C: {mw_seqC:.2f} g/mol")
```

```python
# 3) DNA & mRNA Protein Synthesis
# Transcribe and display the DNA Sequence A, B, and C into mRNA Sequences.
mRNA_SeqA = seqA.transcribe()
print(f"mRNA Sequence A: {mRNA_SeqA}")
mRNA_SeqB = seqB.transcribe()
print(f"mRNA Sequence B: {mRNA_SeqB}")
mRNA_SeqC = seqC.transcribe()
print(f"mRNA Sequence C: {mRNA_SeqC}\n")

# Translate and display the mRNA Sequence A, B, and C into amino acids sequences. (Clue: for translate, use the mRNA sequences that already transcribed on the first step)
# amino_SeqA = seqA.translate() >>> Ini bisa juga, tapi ikutin soal.
amino_SeqA = mRNA_SeqA.translate()
print(f"Amino Acid Sequence A: {amino_SeqA}")
amino_SeqB = mRNA_SeqB.translate()
print(f"Amino Acid Sequence B: {amino_SeqB}")
amino_SeqC = mRNA_SeqC.translate()
print(f"Amino Acid Sequence C: {amino_SeqC}\n")
```

```python
# 4) Sequence Alignment & Similarities
# Perform local alignment pairwise between Sequence A and Sequence “CGCTACTAGCATGCGCAT” and display the alignments’ score. (Clue: Use localxx)
from Bio.pairwise2 import format_alignment, align
localAlignments = align.localxx(seqA, "CGCTACTAGCATGCGCAT")
for alignment in localAlignments:
    print(format_alignment(*alignment))

# Perform global alignment pairwise between Sequence B and Sequence “CAGTGCGCTAGCAGCTACGTGTG” and display the alignments’ score. (Clue: Use globalxx) 
globalAlignments = align.globalxx(seqB, "CAGTGCGCTAGCAGCTACGTGTG")
for alignment in globalAlignments:
    print(format_alignment(*alignment))

# Find and display the Levenshtein distance between Sequence A and B
from Levenshtein import distance
print(f"Levenshtein Distance between Sequence A and B: {distance(seqA, seqB)}")
```
