```python
from Bio.Seq import Seq
```

```python
seqA = Seq("AGTACACTGGT")
print(f'Sequence A: {seqA}')
```

```python
# Indexing
print(f'5th Sequence: {seqA[5]}')

# Slicing -> ambil subsequence dari index start sampai index end 
print(f'Slice 2-5: {seqA[2:6]}')
print(f'Reversed Sequence: {seqA[::-1]}')   # mengambil sequence dari terakhir sampai pertama kalo angka positif dia lompat2 sesuai angkanya
```

```python
# Merge / Join Sequence
seqB = Seq("CTAGGATC")
seqC = seqA + seqB
print(f'Sequence A + B: {seqC}')
```

```python
# Find Codon / Nucleotide Base in Sequence

# Cara 1: .index()
print(f'.index() = Index of Codon "A": {seqA.index("A")}')

# Cara 2: .find()
print(f'.find() = Index of Codon "C": {seqA.find("C")}')
```

```python
# Find Subsequence
from Bio.SeqUtils import nt_search

result = nt_search(str(seqA), "ACT") # Harus diubah ke string dulu
print(f'Raw Result: {result}')
print(f'Subsequence "{result[0]}" Found in Index: {result[1:]}')
```

```python
# Nuclotide Base Frequency

# Cara 1: Counter
from collections import Counter
counter_A = Counter(seqA)
print(counter_A)

print(counter_A.most_common())  # Urut dari yang paling sering muncul
print(counter_A.keys())         # List base yang ada
print(counter_A.values())       # List frekuensi masing2 base

# Cara 2: .count()
print(f'Count of A: {seqA.count("A")}')
print(f'Count of C: {seqA.count("C")}')
print(f'Count of G: {seqA.count("G")}')
print(f'Count of T: {seqA.count("T")}')
```

```python
# Ploting
import matplotlib.pyplot as plt

# Cara 1: Counter
plt.bar(counter_A.keys(), counter_A.values())
plt.title('Cara #1: Counter')
plt.show()

# Cara 2: .count()
plt.bar(['A', 'C', 'G', 'T'], [seqA.count('A'), seqA.count('C'), seqA.count('G'), seqA.count('T')])
plt.title('Cara #2: .count()')
plt.show()
```
