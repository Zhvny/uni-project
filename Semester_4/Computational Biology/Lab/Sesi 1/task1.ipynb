```python
from Bio.Seq import Seq
```

```python
seqA = Seq("AGCTTGCAGCGTCCGTTAGCTCGAGTCCAGGACGTTAGTCCTGCAGTC")
seqB = Seq("CAGTAAGTTGCCGTTAGCGCGTAGTGCCAGTAAGCGGCTCGTTAGTGG")
```

```python
# Find the length of both sequences

# Sequene A
print(f'Length of Sequence A: {len(seqA)}')

# Sequence B
print(f'Length of Sequence B: {len(seqB)}')
```

```python
# Determine the number of times the codon (CGC) appears in both sequences

# Sequence A
print(f'Codon "CGC" appears in Sequence A: {seqA.count("CGC")} times')

# Sequence B
print(f'Codon "CGC" appears in Sequence B: {seqB.count("CGC")} times')
```

```python
# Locate the first occuranve index position of the sub-sequence (CAGTC) in both sequences
from Bio.SeqUtils import nt_search

# Sequence A
result_A = nt_search(str(seqA), "CAGTC")
print(f'Sub-sequence "{result_A[0]}" found in Sequence A at index position: {result_A[1:]}')

# Sequence B
result_B = nt_search(str(seqB), "CAGTC")
print(f'Sub-sequence "{result_B[0]}" found in Sequence B at index position: {result_B[1:]}')
```

```python
# Combine the first 15 nucleotides from sequence A and the last 10 nucleotides from sequence B into a new sequence (seqC) and print it
seqC = seqA[:15] + seqB[-10:]
print(f'Sequence C: {seqC}')
```

```python
# Reverse the sequence of Sequence C
print(f'Reversed Sequence C: {seqC[::-1]}')
```

```python
# Plot the nucleotide base (A, C, G and T) composition of Sequence A, B and C using a bar chart
import matplotlib.pyplot as plt
from collections import Counter

counter_A = Counter(str(seqA))
counter_B = Counter(str(seqB))
counter_C = Counter(str(seqC))

fig, axs = plt.subplots(1, 3, figsize=(15, 5))

axs[0].bar(counter_A.keys(), counter_A.values())
axs[0].set_title('Sequence A')

axs[1].bar(counter_B.keys(), counter_B.values())
axs[1].set_title('Sequence B')

axs[2].bar(counter_C.keys(), counter_C.values())
axs[2].set_title('Sequence C')

plt.tight_layout()
plt.show()
```
