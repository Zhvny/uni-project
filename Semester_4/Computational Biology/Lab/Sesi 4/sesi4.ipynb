```python
from Bio.Seq import Seq
from Bio import pairwise2
from Bio.pairwise2 import format_alignment as fa
```

```python
seq1 = Seq("ATGTTTCCCGGGGTA")
seq2 = Seq("ATGAAACCCTTTGTA")
```

```python
# Local Alignment
local_align = pairwise2.align.localxx(seq1, seq2)
#print(local_align)

# Buat Print Lebih Rapih
for a in local_align:
    print(fa(*a))
```

```python
# Global Alignment
# Score = Match - Mismatch - open_gap - extend_gap
match = +1
missmatch = -1
open_gap = -0.5
extend_gap = -0.1

# Parameter untuk global alignment:
# xx -> function deafult
# mx -> custom match, missmatch
# ms -> custom match, missmatch, open_gap, extend_gap

global_align = pairwise2.align.globalms(seq1, seq2, match, missmatch, open_gap, extend_gap)

for a in global_align:
    print(fa(*a))
```

```python
# Similarity Analysis
# Hamming
def hamming(s1, s2):
    list = []
    for x, y in zip(s1, s2):
        if x != y:
            list.append((x, y))
    return list

print(hamming(seq1, seq2))
```

```python
# Levenshten
from Levenshtein import distance
print(distance(seq1, seq2))
```

```python
# Dotplot
def dotplot(s1, s2, nonblank = chr (0x25A0)):
    print("|", s1, "|")
    print("-" * len(s1))

    for x in s2:
        line = ''
        for y in s1:
            if x == y:
                line += nonblank
            else:
                line += ' '
        print(x + "|" + line)

dotplot(seq1, seq2)
```
