```python
from Bio.Seq import Seq
from Bio import pairwise2
from Bio.pairwise2 import format_alignment as fa
from Levenshtein import distance
```

```python
seq1 = Seq("AAGCTTAGCTAGAGCAGTCAGCTAGCTAGCT")
seq2 = Seq("TCTGAAGCTAGTGCATAGCTAGCTAGC")
```

```python
# Local Alignment
local_align = pairwise2.align.localxx(seq1, seq2)

for a in local_align:
    print(fa(*a))
```

```python
# Hamming Methods
def hamming(s1, s2):
    list = []
    for x, y in zip(s1, s2):
        if x == y:
            list.append((x, y))
    return list
            
print(hamming(seq1, seq2))
```

```python
# Levenshtein Method
print(distance(seq1, seq2))
```

```python
# Dotplot
def dotplot(s1, s2, nonblank = chr (0x25A0)):
    print("|", s1, "|")
    print("-" * len(s1))

    for x in s2:
        line = ''
        for y in s1:
            if x == y:
                line += nonblank
            else:
                line += ' '
        print(x + "|" + line)

dotplot(seq1, seq2)
```
