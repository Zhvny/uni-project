```python
# Import Library
import Bio
from Bio.Seq import Seq
from Bio import SeqIO
```

```python
# Read Fasta
seqA = SeqIO.read("C:/Users/Zhvn/Documents/Downloads/Kuliah/Semester_4/Computational Biology/Environment/a.fasta", "fasta")
seqB = SeqIO.read("C:/Users/Zhvn/Documents/Downloads/Kuliah/Semester_4/Computational Biology/Environment/b.fasta", "fasta")

# Langsung ambil sequence
seqA = seqA.seq
seqB = seqB.seq

# Print Sequence
print("Sequence A : ", seqA)
print("Sequence B : ", seqB)
```

```python
# GC and AT using Biopython
from Bio.SeqUtils import gc_fraction

gc_seqA = gc_fraction(seqA)

print(f"GC Content of Sequence A : {gc_seqA * 100}%")
print(f"AT Content of Sequence A : {(1 - gc_seqA) * 100}%")
```

```python
# Melting Temperature
from Bio.SeqUtils import MeltingTemp as mt

# Method 1: Wallace Rule
MeltingTemp_seqA_wallace = mt.Tm_Wallace(seqA)
print(f"Melting Temperature of Sequence A (Wallace Rule) : {MeltingTemp_seqA_wallace}°C")
MeltingTemp_seqB_wallace = mt.Tm_Wallace(seqB)
print(f"Melting Temperature of Sequence B (Wallace Rule) : {MeltingTemp_seqB_wallace}°C\n")

# Method 2: GC Content
MeltingTemp_seqA_gc = mt.Tm_GC(seqA)
print(f"Melting Temperature of Sequence A (GC Content Method) : {MeltingTemp_seqA_gc}°C")
MeltingTemp_seqB_gc = mt.Tm_GC(seqB)
print(f"Melting Temperature of Sequence B (GC Content Method) : {MeltingTemp_seqB_gc}\n")

# Method 3: Nearest Neighbor
MeltingTemp_seqA_nn = mt.Tm_NN(seqA)
print(f"Melting Temperature of Sequence A (Nearest Neighbor Method) : {MeltingTemp_seqA_nn}°C")
MeltingTemp_seqB_nn = mt.Tm_NN(seqB)
print(f"Melting Temperature of Sequence B (Nearest Neighbor Method) : {MeltingTemp_seqB_nn}\n")
```

```python
# Molecular Weight
from Bio.SeqUtils import molecular_weight

mol_weight_seqA = molecular_weight(seqA)
mol_weight_seqB = molecular_weight(seqB)

print(f"Molecular Weight of Sequence A : {mol_weight_seqA} g/mol")
print(f"Molecular Weight of Sequence B : {mol_weight_seqB} g/mol")
```

```python
# Finding Nucleotide Base Frequency
from collections import Counter

seqA_freq = Counter(seqA)
seqB_freq = Counter(seqB)

print(f"Nucleotide Base Frequency of Sequence A : {seqA_freq}")
print(f"Nucleotide Base Frequency of Sequence B : {seqB_freq}")
```

```python
# Plotting
import matplotlib.pyplot as plt

plt.title("Nucleotide Base Frequency of Sequence A")
plt.bar(seqA_freq.keys(), seqA_freq.values())
plt.show()

plt.title("Nucleotide Base Frequency of Sequence B")
plt.bar(seqB_freq.keys(), seqB_freq.values())
plt.show()
```
