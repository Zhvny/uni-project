```python
import Bio
from Bio.Seq import Seq
from Bio import SeqIO
```

```python
seqA = "AGCTTGCAGCGTCCGTTAGCTCGAGTCCAGGACGTTAGTCCTGCAGTC"
seqB = "CAGTAAGTTGCCGTTAGCGCGTAGTGCCAGTAAGCGGCTCGTTAGTGG"
```

```python
# Determine the molecular weight of both 
from Bio.SeqUtils import molecular_weight

MolecularWeightSeqA = molecular_weight(seqA)
MolecularWeightSeqB = molecular_weight(seqB)

print(f"Molecular Weight of Sequence A : {MolecularWeightSeqA} g/mol")
print(f"Molecular Weight of Sequence B : {MolecularWeightSeqB} g/mol")
```

```python
# Refer to the Wallace rule to calculate the melting point of sequence A and sequence B and plot them using Matplotlib 
import matplotlib.pyplot as plt

plt.title("Molecular Weight of Sequence A and Sequence B")
plt.bar(["Sequence A", "Sequence B"], [MolecularWeightSeqA, MolecularWeightSeqB])
plt.show()
```

```python
from Bio.SeqUtils import gc_fraction

gc_seqA = gc_fraction(seqA)
gc_seqB = gc_fraction(seqB)

# Calculate the GC content of sequence A and sequence B
print(f"GC Content of Sequence A : {gc_seqA * 100}%")
print(f"GC Content of Sequence B : {gc_seqB * 100}%\n")

# Calculate the AT content of sequence A and sequence B
print(f"AT Content of Sequence A : {(1 - gc_seqA) * 100}%")
print(f"AT Content of Sequence B : {(1 - gc_seqB) * 100}%")
```

```python
# Plot the GC and AT content of sequence A and sequence B using Matplotlib library
import matplotlib.pyplot as plt

plt.title("GC and AT Content of Sequence A")
plt.bar(["GC Content", "AT Content"], [gc_seqA * 100, (1 - gc_seqA) * 100])
plt.show()

plt.title("GC and AT Content of Sequence B")
plt.bar(["GC Content", "AT Content"], [gc_seqB * 100, (1 - gc_seqB) * 100])
plt.show()
```
