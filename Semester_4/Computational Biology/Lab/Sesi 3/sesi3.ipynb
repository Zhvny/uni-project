```python
from Bio.Seq import Seq
from Bio.SeqUtils import seq1, seq3

from Bio.Data import CodonTable

import py3Dmol
```

```python
# Protein Syntesis

seq = Seq("ATGATGGCCTATAATATGTAG") # strand sense 5'-3'
# DNA
antisense_seq = seq.reverse_complement() # antisense 3'-5'
print(antisense_seq) 
print(antisense_seq[::-1]) # reversed antisense 5'-3'

mRNA = seq.transcribe() # mRNA 5'-3'
print(mRNA)

amino_acids = mRNA.translate() # asam amino
print(amino_acids) # * = stop codon

custom_amino_acids = mRNA.translate(stop_symbol="NIGGA") # ganti symbol stop
print(custom_amino_acids)

print(seq3(amino_acids)) # proteins

print(seq1(seq3(amino_acids))) # proteins tp 1 huruf
```
